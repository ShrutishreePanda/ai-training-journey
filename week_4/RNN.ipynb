```python
import tensorflow as tf

from keras.models import Sequential

from keras.layers import (
    Embedding,
    SimpleRNN,
    Dense
)

from keras.preprocessing.sequence import pad_sequences
```

```python
(X_train,y_train),(X_test,y_test)=(
    tf.keras.datasets.imdb.load_data(
        num_words=10000
    )
)
```

```python
X_train = pad_sequences(
    X_train,
    maxlen=200
)

X_test = pad_sequences(
    X_test,
    maxlen=200
)
```

```python
model = Sequential([

    Embedding(
        input_dim=10000,
        output_dim=32,
        input_length=200
    ),

    SimpleRNN(
        64
    ),

    Dense(
        1,
        activation="sigmoid"
    )
])
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

```python
loss,acc = model.evaluate(
    X_test,
    y_test
)

print(
    "Test Accuracy:",
    acc
)
```
