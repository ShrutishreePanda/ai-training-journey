```python
import numpy as np
```

```python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])
```

```python
np.random.seed(42)

weights_input_hidden = np.random.randn(
    2,
    4
)

weights_hidden_output = np.random.randn(
    4,
    1
)
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_derivative(x):
    return x*(1-x)
```

```python
learning_rate = 0.1

epochs = 10000
```

```python
for epoch in range(epochs):

    hidden_input = np.dot(
        X,
        weights_input_hidden
    )

    hidden_output = sigmoid(
        hidden_input
    )

    final_input = np.dot(
        hidden_output,
        weights_hidden_output
    )

    final_output = sigmoid(
        final_input
    )

    error = y - final_output

    d_output = (
        error *
        sigmoid_derivative(
            final_output
        )
    )

    hidden_error = np.dot(
        d_output,
        weights_hidden_output.T
    )

    d_hidden = (
        hidden_error *
        sigmoid_derivative(
            hidden_output
        )
    )

    weights_hidden_output += (
        hidden_output.T.dot(
            d_output
        ) * learning_rate
    )

    weights_input_hidden += (
        X.T.dot(
            d_hidden
        ) * learning_rate
    )

    if epoch % 1000 == 0:
        print(
            f"Epoch {epoch}"
        )
```
