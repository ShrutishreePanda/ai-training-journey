```python
##Neuron
# Inputs

x1 = 2
x2 = 3

# Weights

w1 = 0.5
w2 = 0.8

# Bias

b = 0.2

# Weighted Sum

z = (x1 * w1) + (x2 * w2) + b

print("Weighted Sum:", z)
```

```python
#Sigmoid function

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

output = sigmoid(z)

print("Neuron Output:", output)
```

```python
#Activation function

x = np.linspace(-10, 10, 100)
```

```python
#Sigmoid

sigmoid_values = 1 / (1 + np.exp(-x))

plt.figure(figsize=(8,4))
plt.plot(x, sigmoid_values)
plt.title("Sigmoid")
plt.grid()
plt.show()
```

```python
#Tanh

tanh_values = np.tanh(x)

plt.figure(figsize=(8,4))
plt.plot(x, tanh_values)
plt.title("Tanh")
plt.grid()
plt.show()
```

```python
#ReLU

relu_values = np.maximum(0, x)

plt.figure(figsize=(8,4))
plt.plot(x, relu_values)
plt.title("ReLU")
plt.grid()
plt.show()
```

```python
class Neuron:

    def __init__(self, weights, bias):

        self.weights = weights
        self.bias = bias

    def forward(self, inputs):

        z = np.dot(inputs, self.weights) + self.bias

        return sigmoid(z)
```

```python
weights = np.array([0.5, 0.8])

bias = 0.2

neuron = Neuron(weights, bias)

sample = np.array([2,3])

prediction = neuron.forward(sample)

print(prediction)
```

```python
class DenseLayer:

    def __init__(self, input_size, output_size):

        self.weights = np.random.randn(
            input_size,
            output_size
        )

        self.bias = np.zeros(output_size)

    def forward(self, x):

        z = np.dot(x, self.weights) + self.bias

        return sigmoid(z)
```

```python
layer = DenseLayer(
    input_size=2,
    output_size=3
)

sample = np.array([[2,3]])

output = layer.forward(sample)

print(output)
```
